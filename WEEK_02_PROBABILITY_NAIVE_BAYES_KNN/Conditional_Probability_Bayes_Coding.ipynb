# Conditional Probability & Bayes Coding

```python
# importing libraries
import numpy as np
import pandas as pd
```

```python
# creating dataset
data = {
    "Gender": [
        "Female", "Female", "Female", "Female", "Female",
        "Male", "Male", "Male", "Male", "Male"
    ],
    "Passed": [
        "Yes", "Yes", "Yes", "No", "No",
        "Yes", "Yes", "No", "No", "No"
    ]
}

df = pd.DataFrame(data)

df
```

### Marginal Probability

```python
p_female=(df['Gender']=='Female').mean()
p_female
```

```python
p_female_and_passed=(
    (df['Gender']=='Female')&
    (df['Passed']=='Yes')
).mean()
p_female_and_passed
```

```python
pd.crosstab(df['Gender'],df['Passed'],normalize=True)
```

### Conditional Probability

```python
female_students = df[df["Gender"] == "Female"]

female_students
```

```python
# using its formula
p_female_and_passed = (
    (df["Gender"] == "Female") &
    (df["Passed"] == "Yes")
).mean()

p_female = (
    df["Gender"] == "Female"
).mean()

p_passed_given_female = p_female_and_passed / p_female

p_passed_given_female
```

```python
pd.crosstab(
    df['Gender'],
    df['Passed'],
    normalize='index'
)
```

### Contingency Table

### Bayes Theorem

```python
def bayes_theorem(p_a, p_b_given_a, p_b_given_not_a):

    p_b = (
        p_b_given_a * p_a
        +
        p_b_given_not_a * (1 - p_a)
    )

    p_a_given_b = (
        p_b_given_a * p_a
    ) / p_b

    return p_a_given_b
```

```python
result = bayes_theorem(
    p_a=0.01,
    p_b_given_a=0.90,
    p_b_given_not_a=0.05
)

result
```
