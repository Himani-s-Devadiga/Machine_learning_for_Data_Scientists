```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import norm, skew
```

```python
# Create a normal distributionmean = 50
mean=50
std = 10
data = np.random.normal(mean, std, 1000)
data[:10]

print("Mean:", np.mean(data))
print("Standard Deviation:", np.std(data))
```

```python
# Plot the distribution
sns.histplot(data, kde=True)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Normal Distribution")

plt.show()
```

```python
# Understand the parameters
data_1 = np.random.normal(50, 10, 1000)
data_2 = np.random.normal(80, 10, 1000)

plt.hist(data_1, alpha=0.5, label="Mean = 50")
plt.hist(data_2, alpha=0.5, label="Mean = 80")

plt.legend()
plt.show()
```

```python
data_1 = np.random.normal(50, 5, 1000)
data_2 = np.random.normal(50, 15, 1000)

plt.hist(data_1, alpha=0.5, label="SD = 5")
plt.hist(data_2, alpha=0.5, label="SD = 15")

plt.legend()
plt.show()
```

```python
# Standard Normal Variate — Z-score
x = 70
mean = 50
std = 10

z = (x - mean) / std

print("Z-score:", z)
```

```python
# Convert the entire dataset to Z-scores
z_scores = (data - np.mean(data)) / np.std(data)

print(z_scores[:10])
```

```python
print("Mean:", np.mean(z_scores))
print("Standard deviation:", np.std(z_scores))

### This is the connection between today's statistics and your later
# preprocessing topic.
```

```python
# Empirical Rule — 68–95–99.7
mean = np.mean(data)
std = np.std(data)

within_1 = ((data >= mean - std) & (data <= mean + std)).mean()
within_2 = ((data >= mean - 2*std) & (data <= mean + 2*std)).mean()
within_3 = ((data >= mean - 3*std) & (data <= mean + 3*std)).mean()

print("Within 1 SD:", within_1)
print("Within 2 SD:", within_2)
print("Within 3 SD:", within_3)
```

```python
# Probability / area under the curve
mean = 50
std = 10

probability = norm.cdf(60, mean, std)

print(probability)
```

```python
# Probability between two values
probability = norm.cdf(60, mean, std) - norm.cdf(40, mean, std)

print(probability)
```

```python
# CDF
x = np.linspace(20, 80, 1000)
cdf = norm.cdf(x, mean, std)
plt.plot(x, cdf)
plt.xlabel("Value")
plt.ylabel("Cumulative Probability")
plt.title("CDF of Normal Distribution")

plt.show()
```

```python
# Skewness
## Normal distribution
normal_data = np.random.normal(50, 10, 1000)

print("Skewness:", skew(normal_data))
```

```python
right_skewed = np.random.exponential(scale=2, size=1000)

print("Skewness:", skew(right_skewed))
```

```python
left_skewed = -np.random.exponential(scale=2, size=1000)

print("Skewness:", skew(left_skewed))
```

```python
# One small Data Science example
marks = np.random.normal(65, 10, 1000)

print("Mean:", np.mean(marks))
print("Standard deviation:", np.std(marks))
print("Skewness:", skew(marks))
```

```python
sns.histplot(marks, kde=True)

plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Distribution of Student Marks")

plt.show()
```

```python
probability = norm.cdf(80, np.mean(marks), np.std(marks))

print("Probability of scoring below 80:", probability)
```
