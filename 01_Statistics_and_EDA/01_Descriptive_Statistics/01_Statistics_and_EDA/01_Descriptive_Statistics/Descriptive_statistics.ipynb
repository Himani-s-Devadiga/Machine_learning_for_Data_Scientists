# Descriptive statistics

```python
# Import libraries + create dataset
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from statistics import mean,median,multimode,mode
```

```python
data = [10, 12, 15, 15, 18, 20, 22, 25, 30, 35]

data
```

```python
# Mean
sum(data) / len(data)
```

```python
#  Median
np.median(data)
```

```python
#  Mode
mode(data)
```

```python
# Weighted Mean

marks = [80, 70, 90]
weights = [0.2, 0.3, 0.5]
weighted_mean = np.average(marks, weights=weights)

weighted_mean
```

```python
# Trimmed Mean

from scipy.stats import trim_mean
data_with_outliers = [10, 12, 15, 18, 20, 22, 25, 30, 100]
trim_mean(data_with_outliers, proportiontocut=0.1)
```

```python
# Range

max(data) - min(data)
```

```python
np.ptp(data)
```

```python
# Variance

#Population variance
np.var(data)
```

```python
# Sample variance
np.var(data, ddof=1)

# Remember:

# ddof=0 → population variance
# ddof=1 → sample variance
```

```python
#  Standard Deviation
# Population SD
np.std(data)
```

```python
# Sample SD
np.std(data, ddof=1)
```

```python
# Coefficient of Variation

mean_value = np.mean(data)
std_value = np.std(data)

cv = (std_value / mean_value) * 100

cv
```

```python
# Create a DataFrame

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Age": [20, 21, 22, 20, 23, 21, 22, 24, 20, 25],
    "Marks": [78, 85, 92, 67, 88, 76, 95, 82, 70, 90],
    "Department": [
        "CSE", "ISE", "CSE", "ECE", "CSE",
        "ISE", "ECE", "CSE", "ECE", "ISE"
    ]
})

df
```

```python
# Basic Data Understanding
df.head()
```

```python
df.tail()
```

```python
df.describe()
```

```python
# Frequency Distribution
df["Department"].value_counts()
```

```python
# Cumulative Frequency
frequency_table = df["Department"].value_counts().reset_index()

frequency_table.columns = ["Department", "Frequency"]

frequency_table
```

```python
frequency_table["Cumulative_Frequency"] = frequency_table["Frequency"].cumsum()

frequency_table
```

```python
# Scatter Plot — Numerical vs Numerical
plt.scatter(df["Age"], df["Marks"])

plt.xlabel("Age")
plt.ylabel("Marks")
plt.title("Age vs Marks")
plt.show()
```

```python
# Categorical vs Numerical
df.boxplot(column="Marks", by="Department")

plt.xlabel("Department")
plt.ylabel("Marks")
plt.title("Marks by Department")
plt.suptitle("")

plt.show()
```

```python
# Crosstab — Categorical vs Categorical
df["Gender"] = [
    "F", "M", "F", "M", "F",
    "F", "M", "F", "M", "F"
]
pd.crosstab(df["Department"], df["Gender"])
```
