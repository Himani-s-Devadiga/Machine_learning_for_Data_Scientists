# Univariate Analysis

```python
# Missing values
df.isnull().sum()

# Percentage:
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_percentage.sort_values(ascending=False)

# Visualize:
plt.figure(figsize=(10,5))
sns.heatmap(df.isnull(), cbar=False)
plt.title("Missing Values in Netflix Dataset")
plt.show()
```

### Key Finding:
Missing values are not evenly distributed across columns. Director, cast and country contain substantially more missing information than the core fields such as title, type and release year.

# Content type

```python
df["type"].value_counts()
```

```python
# Percentage:
df["type"].value_counts(normalize=True) * 100
```

# Visualization

```python
sns.countplot(data=df, x="type")

plt.title("Distribution of Netflix Content Types")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.show()
```

### Key Finding:
Movies make up the majority of titles in the dataset, while TV Shows represent a smaller share.

# Release year distribution

```python
df["release_year"].describe()
```

# Histogram

```python
plt.figure(figsize=(10,5))

sns.histplot(data=df, x="release_year", bins=30)

plt.title("Distribution of Release Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

### Key finding
You should see that Netflix's catalog is heavily concentrated toward more recent release years, rather than being evenly distributed across the entire historical range.

# Content released over time

```python
yearly_content = df.groupby("release_year").size()

plt.figure(figsize=(12,5))

yearly_content.plot(kind="line")

plt.title("Number of Netflix Titles by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

# Movie vs TV Show over time

```python
type_by_year = df.groupby(["release_year", "type"]).size().unstack()

type_by_year.tail()
```

```python
type_by_year.plot(figsize=(12,6))

plt.title("Movies vs TV Shows by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

### Key finding

Movies generally have a larger presence than TV Shows across the dataset, although the relative distribution varies across release years.

```python
# Ratings
df["rating"].value_counts()
```

```python
# Top ratings:

df["rating"].value_counts().head(10)
```

```python
# Plot:

plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    y="rating",
    order=df["rating"].value_counts().index
)

plt.title("Netflix Content by Rating")
plt.xlabel("Number of Titles")
plt.ylabel("Rating")
plt.show()
```

# Countries

```python
df["country"].value_counts().head(10)
```

```python
# Plot:

top_countries = df["country"].value_counts().head(10)

top_countries.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Top 10 Countries by Netflix Titles")
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)

plt.show()
```

### Key finding

The United States is the dominant country in the raw country field.

But there is an important problem:

United States, India
United States, Canada
India, United Kingdom

are treated as individual strings.

So a simple value_counts() doesn't perfectly measure country contribution.

This is actually a very good EDA insight:

The country column contains multiple countries in some rows, so simple frequency counting can underestimate the contribution of individual countries.


```python
# Top directors
df["director"].value_counts().head(10)

# Plot:

df["director"].value_counts().head(10).plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Top 10 Directors by Number of Netflix Titles")
plt.xlabel("Director")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)

plt.show()
```

### Key Finding:

Because director has many missing values, this analysis only considers records where a director is listed.
